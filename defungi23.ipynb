```python
# ============================================================
# Q1-READY MICROSCOPIC FUNGI CLASSIFICATION PIPELINE
# EfficientNetV2-S + CBAM + Advanced Preprocessing
# FULL END-TO-END NOTEBOOK
# ============================================================

# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

!pip install timm opencv-python-headless seaborn scikit-learn -q

# ============================================================
# 2. IMPORTS
# ============================================================

import os
import cv2
import random
import shutil
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torchvision import transforms

import timm

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    cohen_kappa_score,
    matthews_corrcoef,
    roc_auc_score,
    roc_curve,
    auc,
    f1_score,
)

from sklearn.preprocessing import label_binarize

warnings.filterwarnings("ignore")

# ============================================================
# 3. CONFIGURATION
# ============================================================

ORIGINAL_DATASET = "/kaggle/input/datasets/joebeachcapital/defungi"

SPLIT_DATASET = "/kaggle/working/fungi_split_dataset"

SAVE_DIR = "/kaggle/working"

IMG_SIZE = 288
BATCH_SIZE = 32
EPOCHS = 40

LR = 2e-4
WEIGHT_DECAY = 1e-4

PATIENCE = 10
NUM_WORKERS = 2
SEED = 42

TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

# ============================================================
# 4. SEED
# ============================================================

def seed_everything(seed=42):

    random.seed(seed)
    np.random.seed(seed)

    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True

seed_everything(SEED)

# ============================================================
# 5. DEVICE
# ============================================================

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("="*60)
print("DEVICE:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

print("="*60)

# ============================================================
# 6. DATASET SPLITTING
# ============================================================

print("\nSplitting Dataset...\n")

classes = ['H1', 'H2', 'H3', 'H5', 'H6']

for split in ['train', 'val', 'test']:
    for cls in classes:
        os.makedirs(os.path.join(SPLIT_DATASET, split, cls), exist_ok=True)

before_stats = []
after_stats = []

for cls in classes:

    class_path = os.path.join(ORIGINAL_DATASET, cls)

    images = [
        img for img in os.listdir(class_path)
        if img.lower().endswith(('.png', '.jpg', '.jpeg'))
    ]

    before_stats.append([cls, len(images)])

    train_imgs, temp_imgs = train_test_split(
        images,
        test_size=(1 - TRAIN_RATIO),
        random_state=SEED,
        shuffle=True
    )

    val_imgs, test_imgs = train_test_split(
        temp_imgs,
        test_size=0.5,
        random_state=SEED,
        shuffle=True
    )

    for img in train_imgs:
        shutil.copy2(
            os.path.join(class_path, img),
            os.path.join(SPLIT_DATASET, 'train', cls, img)
        )

    for img in val_imgs:
        shutil.copy2(
            os.path.join(class_path, img),
            os.path.join(SPLIT_DATASET, 'val', cls, img)
        )

    for img in test_imgs:
        shutil.copy2(
            os.path.join(class_path, img),
            os.path.join(SPLIT_DATASET, 'test', cls, img)
        )

    after_stats.append([
        cls,
        len(images),
        len(train_imgs),
        len(val_imgs),
        len(test_imgs)
    ])

before_df = pd.DataFrame(
    before_stats,
    columns=["Class", "Original"]
)

after_df = pd.DataFrame(
    after_stats,
    columns=["Class", "Original", "Train", "Val", "Test"]
)

print("\nBEFORE SPLIT\n")
print(before_df)

print("\nAFTER SPLIT\n")
print(after_df)

before_df.to_csv(
    os.path.join(SAVE_DIR, "before_split_statistics.csv"),
    index=False
)

after_df.to_csv(
    os.path.join(SAVE_DIR, "after_split_statistics.csv"),
    index=False
)

# ============================================================
# 7. PREPROCESSING PIPELINE
# ============================================================

MORPH_KERNEL = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE,
    (3, 3)
)

def step1_median_denoise(img):
    return cv2.medianBlur(img, 3)

def step2_morphological_clean(img):

    opened = cv2.morphologyEx(
        img,
        cv2.MORPH_OPEN,
        MORPH_KERNEL
    )

    closed = cv2.morphologyEx(
        opened,
        cv2.MORPH_CLOSE,
        MORPH_KERNEL
    )

    return closed

def step3_clahe(img):

    lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)

    l, a, b = cv2.split(lab)

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    l_eq = clahe.apply(l)

    merged = cv2.merge([l_eq, a, b])

    return cv2.cvtColor(
        merged,
        cv2.COLOR_LAB2BGR
    )

def step4_stain_normalize(img):

    target_mean = np.array(
        [142.0, 130.0, 127.0],
        dtype=np.float32
    )

    target_std = np.array(
        [36.0, 12.0, 11.0],
        dtype=np.float32
    )

    lab = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2LAB
    ).astype(np.float32)

    for i in range(3):

        mu = lab[:, :, i].mean()

        sig = lab[:, :, i].std() + 1e-6

        lab[:, :, i] = (
            (lab[:, :, i] - mu)
            / sig
            * target_std[i]
            + target_mean[i]
        )

    lab = np.clip(lab, 0, 255).astype(np.uint8)

    return cv2.cvtColor(
        lab,
        cv2.COLOR_LAB2BGR
    )

def step5_unsharp_mask(img, strength=0.5):

    blurred = cv2.GaussianBlur(
        img,
        (0, 0),
        sigmaX=1.5
    )

    sharp = cv2.addWeighted(
        img,
        1 + strength,
        blurred,
        -strength,
        0
    )

    return np.clip(sharp, 0, 255).astype(np.uint8)

def preprocess_image(pil_img):

    img = cv2.cvtColor(
        np.array(pil_img),
        cv2.COLOR_RGB2BGR
    )

    img = step1_median_denoise(img)

    img = step2_morphological_clean(img)

    img = step3_clahe(img)

    img = step4_stain_normalize(img)

    img = step5_unsharp_mask(img)

    return Image.fromarray(
        cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    )

# ============================================================
# 8. COMPUTE DATASET NORMALIZATION
# ============================================================

print("\nComputing Dataset Mean and Std...\n")

TRAIN_DIR = os.path.join(SPLIT_DATASET, "train")

CLASS_NAMES = sorted(os.listdir(TRAIN_DIR))

def compute_dataset_stats():

    all_paths = []

    for c in CLASS_NAMES:

        folder = os.path.join(TRAIN_DIR, c)

        for f in os.listdir(folder):

            all_paths.append(
                os.path.join(folder, f)
            )

    random.shuffle(all_paths)

    paths = all_paths[:2000]

    tfms = transforms.Compose([
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.ToTensor()
    ])

    mean = torch.zeros(3)
    std = torch.zeros(3)

    n = 0

    for path in tqdm(paths):

        try:

            img = Image.open(path).convert("RGB")

            img = preprocess_image(img)

            t = tfms(img)

            mean += t.mean(dim=[1,2])

            std += t.std(dim=[1,2])

            n += 1

        except:
            pass

    mean /= n
    std /= n

    return mean.tolist(), std.tolist()

MEAN, STD = compute_dataset_stats()

print("\nMEAN:", MEAN)
print("STD :", STD)

# ============================================================
# 9. TRANSFORMS
# ============================================================

train_tfms = transforms.Compose([

    transforms.Resize((IMG_SIZE, IMG_SIZE)),

    transforms.RandomHorizontalFlip(),

    transforms.RandomVerticalFlip(),

    transforms.RandomRotation(20),

    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),

    transforms.ToTensor(),

    transforms.Normalize(MEAN, STD)
])

val_tfms = transforms.Compose([

    transforms.Resize((IMG_SIZE, IMG_SIZE)),

    transforms.ToTensor(),

    transforms.Normalize(MEAN, STD)
])

# ============================================================
# 10. DATAFRAME
# ============================================================

label_map = {
    c:i for i,c in enumerate(CLASS_NAMES)
}

NUM_CLASSES = len(CLASS_NAMES)

def make_df(split):

    rows = []

    for c in CLASS_NAMES:

        folder = os.path.join(
            SPLIT_DATASET,
            split,
            c
        )

        for img in os.listdir(folder):

            rows.append({
                "path": os.path.join(folder, img),
                "label": c
            })

    return pd.DataFrame(rows)

train_df = make_df("train")
val_df = make_df("val")
test_df = make_df("test")

print("\nTrain:", len(train_df))
print("Val:", len(val_df))
print("Test:", len(test_df))

# ============================================================
# 11. DATASET CLASS
# ============================================================

class FungiDataset(Dataset):

    def __init__(self, df, tfms):

        self.df = df.reset_index(drop=True)

        self.tfms = tfms

    def __len__(self):

        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        img = Image.open(
            row["path"]
        ).convert("RGB")

        img = preprocess_image(img)

        img = self.tfms(img)

        label = label_map[row["label"]]

        return img, label

train_ds = FungiDataset(train_df, train_tfms)

val_ds = FungiDataset(val_df, val_tfms)

test_ds = FungiDataset(test_df, val_tfms)

# ============================================================
# 12. WEIGHTED SAMPLER
# ============================================================

counts = train_df["label"].value_counts()

sample_weights = train_df["label"].map(
    lambda x: 1.0 / counts[x]
).values

sampler = WeightedRandomSampler(
    torch.DoubleTensor(sample_weights),
    len(sample_weights),
    replacement=True
)

# ============================================================
# 13. DATALOADER
# ============================================================

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    sampler=sampler,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

# ============================================================
# 14. CBAM
# ============================================================

class ChannelAttention(nn.Module):

    def __init__(self, channels, reduction=16):

        super().__init__()

        self.avg_pool = nn.AdaptiveAvgPool2d(1)

        self.max_pool = nn.AdaptiveMaxPool2d(1)

        self.fc = nn.Sequential(

            nn.Conv2d(
                channels,
                channels // reduction,
                1,
                bias=False
            ),

            nn.ReLU(),

            nn.Conv2d(
                channels // reduction,
                channels,
                1,
                bias=False
            )
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        avg_out = self.fc(self.avg_pool(x))

        max_out = self.fc(self.max_pool(x))

        out = avg_out + max_out

        return x * self.sigmoid(out)

class SpatialAttention(nn.Module):

    def __init__(self):

        super().__init__()

        self.conv = nn.Conv2d(
            2,
            1,
            kernel_size=7,
            padding=3
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        avg_out = torch.mean(
            x,
            dim=1,
            keepdim=True
        )

        max_out, _ = torch.max(
            x,
            dim=1,
            keepdim=True
        )

        x_cat = torch.cat(
            [avg_out, max_out],
            dim=1
        )

        return x * self.sigmoid(
            self.conv(x_cat)
        )

class CBAM(nn.Module):

    def __init__(self, channels):

        super().__init__()

        self.ca = ChannelAttention(channels)

        self.sa = SpatialAttention()

    def forward(self, x):

        x = self.ca(x)

        x = self.sa(x)

        return x

# ============================================================
# 15. MODEL
# ============================================================

class FungiModel(nn.Module):

    def __init__(self, num_classes):

        super().__init__()

        self.backbone = timm.create_model(
            "tf_efficientnetv2_s",
            pretrained=True,
            num_classes=0,
            global_pool=""
        )

        features = self.backbone.num_features

        self.cbam = CBAM(features)

        self.pool = nn.AdaptiveAvgPool2d(1)

        self.classifier = nn.Sequential(

            nn.Flatten(),

            nn.Linear(features, 512),

            nn.BatchNorm1d(512),

            nn.GELU(),

            nn.Dropout(0.3),

            nn.Linear(512, num_classes)
        )

    def forward(self, x):

        x = self.backbone(x)

        x = self.cbam(x)

        x = self.pool(x)

        x = self.classifier(x)

        return x

model = FungiModel(NUM_CLASSES).to(device)

# ============================================================
# 16. LOSS / OPTIMIZER / SCHEDULER
# ============================================================

criterion = nn.CrossEntropyLoss(
    label_smoothing=0.1
)

optimizer = optim.AdamW(
    model.parameters(),
    lr=LR,
    weight_decay=WEIGHT_DECAY
)

scheduler = optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=EPOCHS
)

scaler = torch.amp.GradScaler("cuda")

# ============================================================
# 17. TRAIN FUNCTION
# ============================================================

def train_epoch():

    model.train()

    total_loss = 0

    correct = 0

    total = 0

    loop = tqdm(train_loader)

    for images, labels in loop:

        images = images.to(device)

        labels = labels.to(device)

        optimizer.zero_grad()

        with torch.amp.autocast("cuda"):

            outputs = model(images)

            loss = criterion(outputs, labels)

        scaler.scale(loss).backward()

        scaler.step(optimizer)

        scaler.update()

        total_loss += loss.item()

        _, preds = torch.max(outputs, 1)

        correct += (preds == labels).sum().item()

        total += labels.size(0)

        loop.set_description(
            f"Loss: {loss.item():.4f}"
        )

    acc = 100 * correct / total

    return total_loss / len(train_loader), acc

# ============================================================
# 18. VALIDATION
# ============================================================

@torch.no_grad()
def validate():

    model.eval()

    correct = 0

    total = 0

    val_loss = 0

    for images, labels in val_loader:

        images = images.to(device)

        labels = labels.to(device)

        outputs = model(images)

        loss = criterion(outputs, labels)

        val_loss += loss.item()

        _, preds = torch.max(outputs, 1)

        correct += (preds == labels).sum().item()

        total += labels.size(0)

    acc = 100 * correct / total

    return val_loss / len(val_loader), acc

# ============================================================
# 19. TRAINING
# ============================================================

best_acc = 0

patience_counter = 0

train_losses = []

val_losses = []

train_accs = []

val_accs = []

print("\n" + "="*60)
print("TRAINING STARTED")
print("="*60)

for epoch in range(EPOCHS):

    print(f"\nEpoch [{epoch+1}/{EPOCHS}]")

    train_loss, train_acc = train_epoch()

    val_loss, val_acc = validate()

    scheduler.step()

    train_losses.append(train_loss)

    val_losses.append(val_loss)

    train_accs.append(train_acc)

    val_accs.append(val_acc)

    print(f"Train Loss : {train_loss:.4f}")

    print(f"Val Loss   : {val_loss:.4f}")

    print(f"Train Acc  : {train_acc:.2f}%")

    print(f"Val Acc    : {val_acc:.2f}%")

    if val_acc > best_acc:

        best_acc = val_acc

        torch.save(
            model.state_dict(),
            "best_fungi.pth"
        )

        patience_counter = 0

        print("Saved Best Model")

    else:

        patience_counter += 1

        print(
            f"Patience: {patience_counter}/{PATIENCE}"
        )

    if patience_counter >= PATIENCE:

        print("Early Stopping")

        break

print("\nBest Validation Accuracy:", best_acc)

# ============================================================
# 20. LOAD BEST MODEL
# ============================================================

model.load_state_dict(
    torch.load("best_fungi.pth")
)

# ============================================================
# 21. TESTING
# ============================================================

@torch.no_grad()
def test():

    model.eval()

    y_true = []

    y_pred = []

    y_prob = []

    for images, labels in tqdm(test_loader):

        images = images.to(device)

        outputs = model(images)

        probs = torch.softmax(outputs, dim=1)

        preds = torch.argmax(probs, dim=1)

        y_true.extend(labels.numpy())

        y_pred.extend(preds.cpu().numpy())

        y_prob.extend(probs.cpu().numpy())

    return (
        np.array(y_true),
        np.array(y_pred),
        np.array(y_prob)
    )

y_true, y_pred, y_prob = test()

# ============================================================
# 22. METRICS
# ============================================================

acc = accuracy_score(y_true, y_pred)

f1 = f1_score(
    y_true,
    y_pred,
    average="weighted"
)

kappa = cohen_kappa_score(
    y_true,
    y_pred
)

mcc = matthews_corrcoef(
    y_true,
    y_pred
)

print("\n" + "="*60)
print("FINAL RESULTS")
print("="*60)

print(f"Accuracy : {acc*100:.2f}%")
print(f"F1 Score : {f1:.4f}")
print(f"Kappa    : {kappa:.4f}")
print(f"MCC      : {mcc:.4f}")

print("\nClassification Report:\n")

print(classification_report(
    y_true,
    y_pred,
    target_names=CLASS_NAMES
))

# ============================================================
# 23. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(8,6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=CLASS_NAMES,
    yticklabels=CLASS_NAMES
)

plt.xlabel("Predicted")

plt.ylabel("True")

plt.title("Confusion Matrix")

plt.savefig(
    os.path.join(SAVE_DIR, "confusion_matrix.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 24. ROC CURVE
# ============================================================

y_true_bin = label_binarize(
    y_true,
    classes=list(range(NUM_CLASSES))
)

plt.figure(figsize=(8,6))

for i in range(NUM_CLASSES):

    fpr, tpr, _ = roc_curve(
        y_true_bin[:, i],
        y_prob[:, i]
    )

    roc_auc = auc(fpr, tpr)

    plt.plot(
        fpr,
        tpr,
        label=f"{CLASS_NAMES[i]} (AUC={roc_auc:.3f})"
    )

plt.plot([0,1],[0,1],"k--")

plt.xlabel("False Positive Rate")

plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.legend()

plt.savefig(
    os.path.join(SAVE_DIR, "roc_curve.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 25. SAVE METRICS
# ============================================================

summary = pd.DataFrame([{
    "Accuracy": acc,
    "F1": f1,
    "Kappa": kappa,
    "MCC": mcc
}])

summary.to_csv(
    os.path.join(
        SAVE_DIR,
        "metrics_summary.csv"
    ),
    index=False
)

# ============================================================
# 26. FINISHED
# ============================================================

print("\n" + "="*60)
print("PIPELINE COMPLETE")
print("="*60)

print("\nSaved Files:")

print("before_split_statistics.csv")
print("after_split_statistics.csv")
print("confusion_matrix.png")
print("roc_curve.png")
print("metrics_summary.csv")
print("best_fungi.pth")
```

```python
# ============================================================
# INSERTION & DELETION CURVES
# EXPLAINABILITY EVALUATION FOR GRAD-CAM
# JOURNAL-READY IMPLEMENTATION
# ============================================================

# ============================================================
# IMPORTS
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from tqdm import tqdm

# ============================================================
# DIRECTORY
# ============================================================

EXPLAIN_DIR = os.path.join(
    SAVE_DIR,
    "explainability_curves"
)

os.makedirs(EXPLAIN_DIR, exist_ok=True)

# ============================================================
# HELPER FUNCTION
# ============================================================

def predict_confidence(input_tensor):

    with torch.no_grad():

        output = model(input_tensor)

        prob = torch.softmax(
            output,
            dim=1
        )

        confidence, pred = torch.max(
            prob,
            dim=1
        )

    return confidence.item(), pred.item()

# ============================================================
# SELECT SAMPLE IMAGE
# ============================================================

sample_img, sample_label = test_ds[0]

input_tensor = sample_img.unsqueeze(0).to(device)

# ============================================================
# ORIGINAL PREDICTION
# ============================================================

orig_conf, orig_pred = predict_confidence(
    input_tensor
)

print(f"Original Confidence : {orig_conf:.4f}")

# ============================================================
# GENERATE GRAD-CAM
# ============================================================

grayscale_cam = cam(
    input_tensor=input_tensor
)[0]

# ============================================================
# DENORMALIZE IMAGE
# ============================================================

img_np = inv_norm(sample_img)

img_np = img_np.permute(1,2,0)

img_np = img_np.clamp(0,1).numpy()

# ============================================================
# FLATTEN IMPORTANCE MAP
# ============================================================

importance = grayscale_cam.flatten()

sorted_idx = np.argsort(
    importance
)[::-1]

# ============================================================
# PARAMETERS
# ============================================================

steps = 20

pixels_per_step = len(sorted_idx) // steps

insertion_scores = []

deletion_scores = []

# ============================================================
# INSERTION CURVE
# Start from black image and insert important pixels
# ============================================================

insert_img = np.zeros_like(img_np)

# ============================================================
# DELETION CURVE
# Start from original image and remove important pixels
# ============================================================

delete_img = img_np.copy()

# ============================================================
# RUN CURVES
# ============================================================

for step in tqdm(range(steps)):

    start = step * pixels_per_step

    end = (step + 1) * pixels_per_step

    idxs = sorted_idx[start:end]

    # ========================================================
    # PIXEL LOCATIONS
    # ========================================================

    h = idxs // grayscale_cam.shape[1]

    w = idxs % grayscale_cam.shape[1]

    # ========================================================
    # INSERTION
    # ========================================================

    insert_img[h,w,:] = img_np[h,w,:]

    insert_tensor = transforms.ToTensor()(
        insert_img
    )

    insert_tensor = transforms.Normalize(
        MEAN,
        STD
    )(insert_tensor)

    insert_tensor = insert_tensor.unsqueeze(0).to(device)

    insert_conf, _ = predict_confidence(
        insert_tensor
    )

    insertion_scores.append(insert_conf)

    # ========================================================
    # DELETION
    # ========================================================

    delete_img[h,w,:] = 0

    delete_tensor = transforms.ToTensor()(
        delete_img
    )

    delete_tensor = transforms.Normalize(
        MEAN,
        STD
    )(delete_tensor)

    delete_tensor = delete_tensor.unsqueeze(0).to(device)

    delete_conf, _ = predict_confidence(
        delete_tensor
    )

    deletion_scores.append(delete_conf)

# ============================================================
# X-AXIS
# ============================================================

fractions = np.linspace(
    0,
    1,
    steps
)

# ============================================================
# PLOT INSERTION CURVE
# ============================================================

plt.figure(figsize=(7,5))

plt.plot(
    fractions,
    insertion_scores,
    marker='o',
    linewidth=2,
    label="Insertion"
)

plt.xlabel(
    "Fraction of Inserted Important Pixels"
)

plt.ylabel(
    "Prediction Confidence"
)

plt.title(
    "Insertion Curve"
)

plt.grid(True)

plt.legend()

plt.savefig(
    os.path.join(
        EXPLAIN_DIR,
        "insertion_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# PLOT DELETION CURVE
# ============================================================

plt.figure(figsize=(7,5))

plt.plot(
    fractions,
    deletion_scores,
    marker='o',
    linewidth=2,
    label="Deletion"
)

plt.xlabel(
    "Fraction of Removed Important Pixels"
)

plt.ylabel(
    "Prediction Confidence"
)

plt.title(
    "Deletion Curve"
)

plt.grid(True)

plt.legend()

plt.savefig(
    os.path.join(
        EXPLAIN_DIR,
        "deletion_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# COMBINED CURVE
# ============================================================

plt.figure(figsize=(8,6))

plt.plot(
    fractions,
    insertion_scores,
    marker='o',
    linewidth=2,
    label="Insertion"
)

plt.plot(
    fractions,
    deletion_scores,
    marker='s',
    linewidth=2,
    label="Deletion"
)

plt.xlabel(
    "Fraction of Important Pixels"
)

plt.ylabel(
    "Prediction Confidence"
)

plt.title(
    "Insertion and Deletion Curves"
)

plt.grid(True)

plt.legend()

plt.savefig(
    os.path.join(
        EXPLAIN_DIR,
        "insertion_deletion_curve.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# AREA UNDER CURVES
# ============================================================

insertion_auc = np.trapz(
    insertion_scores,
    fractions
)

deletion_auc = np.trapz(
    deletion_scores,
    fractions
)

print("\n" + "="*60)

print(f"Insertion AUC : {insertion_auc:.4f}")

print(f"Deletion AUC : {deletion_auc:.4f}")

print("="*60)

# ============================================================
# SAVE RESULTS
# ============================================================

curve_df = pd.DataFrame({

    "Fraction": fractions,

    "Insertion": insertion_scores,

    "Deletion": deletion_scores
})

curve_df.to_csv(
    os.path.join(
        EXPLAIN_DIR,
        "insertion_deletion_scores.csv"
    ),
    index=False
)

# ============================================================
# FINAL MESSAGE
# ============================================================

print("\nGenerated Files:")

print("""
1. insertion_curve.png
2. deletion_curve.png
3. insertion_deletion_curve.png
4. insertion_deletion_scores.csv
""")
```

```python
# ============================================================
# Q1-READY MICROSCOPIC FUNGI CLASSIFICATION PIPELINE
# MULTI-MODEL COMPARISON FRAMEWORK
# EfficientNetV2-S + CBAM vs 6 Baseline Models
# ============================================================

# ============================================================
# 1. INSTALL LIBRARIES
# ============================================================

!pip install timm opencv-python-headless seaborn thop -q

# ============================================================
# 2. IMPORTS
# ============================================================

import os
import cv2
import time
import random
import shutil
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader
from torch.utils.data import WeightedRandomSampler

from torchvision import transforms

import timm

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    roc_curve,
    auc,
    precision_recall_curve,
    cohen_kappa_score,
    matthews_corrcoef
)

from sklearn.preprocessing import label_binarize

from thop import profile

warnings.filterwarnings("ignore")

# ============================================================
# 3. CONFIGURATION
# ============================================================

ORIGINAL_DATASET = "/kaggle/input/datasets/joebeachcapital/defungi"

SPLIT_DATASET = "/kaggle/working/fungi_split_dataset"

SAVE_DIR = "/kaggle/working"

IMG_SIZE = 288

BATCH_SIZE = 32

EPOCHS = 20

LR = 2e-4

WEIGHT_DECAY = 1e-4

NUM_WORKERS = 2

PATIENCE = 5

SEED = 42

TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

# ============================================================
# 4. SEED
# ============================================================

def seed_everything(seed=42):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True

seed_everything(SEED)

# ============================================================
# 5. DEVICE
# ============================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("="*60)
print("DEVICE:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

print("="*60)

# ============================================================
# 6. DATASET SPLITTING
# ============================================================

classes = ['H1', 'H2', 'H3', 'H5', 'H6']

for split in ['train', 'val', 'test']:

    for cls in classes:

        os.makedirs(
            os.path.join(
                SPLIT_DATASET,
                split,
                cls
            ),
            exist_ok=True
        )

for cls in classes:

    class_path = os.path.join(
        ORIGINAL_DATASET,
        cls
    )

    images = [

        img for img in os.listdir(class_path)

        if img.lower().endswith(
            ('.png', '.jpg', '.jpeg')
        )
    ]

    train_imgs, temp_imgs = train_test_split(

        images,

        test_size=(1 - TRAIN_RATIO),

        random_state=SEED,

        shuffle=True
    )

    val_imgs, test_imgs = train_test_split(

        temp_imgs,

        test_size=0.5,

        random_state=SEED,

        shuffle=True
    )

    for img in train_imgs:

        shutil.copy2(

            os.path.join(class_path, img),

            os.path.join(
                SPLIT_DATASET,
                'train',
                cls,
                img
            )
        )

    for img in val_imgs:

        shutil.copy2(

            os.path.join(class_path, img),

            os.path.join(
                SPLIT_DATASET,
                'val',
                cls,
                img
            )
        )

    for img in test_imgs:

        shutil.copy2(

            os.path.join(class_path, img),

            os.path.join(
                SPLIT_DATASET,
                'test',
                cls,
                img
            )
        )

print("\nDataset Split Complete")

# ============================================================
# 7. PREPROCESSING
# ============================================================

MORPH_KERNEL = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE,
    (3,3)
)

def step1_median_denoise(img):

    return cv2.medianBlur(img, 3)

def step2_morphological_clean(img):

    opened = cv2.morphologyEx(
        img,
        cv2.MORPH_OPEN,
        MORPH_KERNEL
    )

    closed = cv2.morphologyEx(
        opened,
        cv2.MORPH_CLOSE,
        MORPH_KERNEL
    )

    return closed

def step3_clahe(img):

    lab = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2LAB
    )

    l,a,b = cv2.split(lab)

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8,8)
    )

    l_eq = clahe.apply(l)

    merged = cv2.merge([l_eq,a,b])

    return cv2.cvtColor(
        merged,
        cv2.COLOR_LAB2BGR
    )

def step4_stain_normalize(img):

    target_mean = np.array(
        [142,130,127],
        dtype=np.float32
    )

    target_std = np.array(
        [36,12,11],
        dtype=np.float32
    )

    lab = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2LAB
    ).astype(np.float32)

    for i in range(3):

        mu = lab[:,:,i].mean()

        sig = lab[:,:,i].std() + 1e-6

        lab[:,:,i] = (
            (lab[:,:,i] - mu)
            / sig
            * target_std[i]
            + target_mean[i]
        )

    lab = np.clip(
        lab,
        0,
        255
    ).astype(np.uint8)

    return cv2.cvtColor(
        lab,
        cv2.COLOR_LAB2BGR
    )

def step5_unsharp_mask(img):

    blurred = cv2.GaussianBlur(
        img,
        (0,0),
        sigmaX=1.5
    )

    sharp = cv2.addWeighted(
        img,
        1.5,
        blurred,
        -0.5,
        0
    )

    return np.clip(
        sharp,
        0,
        255
    ).astype(np.uint8)

def preprocess_image(pil_img):

    img = cv2.cvtColor(
        np.array(pil_img),
        cv2.COLOR_RGB2BGR
    )

    img = step1_median_denoise(img)

    img = step2_morphological_clean(img)

    img = step3_clahe(img)

    img = step4_stain_normalize(img)

    img = step5_unsharp_mask(img)

    return Image.fromarray(
        cv2.cvtColor(
            img,
            cv2.COLOR_BGR2RGB
        )
    )

# ============================================================
# 8. DATASET STATS
# ============================================================

TRAIN_DIR = os.path.join(
    SPLIT_DATASET,
    "train"
)

CLASS_NAMES = sorted(
    os.listdir(TRAIN_DIR)
)

label_map = {
    c:i for i,c in enumerate(CLASS_NAMES)
}

NUM_CLASSES = len(CLASS_NAMES)

def compute_dataset_stats():

    all_paths = []

    for c in CLASS_NAMES:

        folder = os.path.join(
            TRAIN_DIR,
            c
        )

        for f in os.listdir(folder):

            all_paths.append(
                os.path.join(folder,f)
            )

    random.shuffle(all_paths)

    paths = all_paths[:1500]

    tfms = transforms.Compose([

        transforms.Resize((IMG_SIZE,IMG_SIZE)),

        transforms.ToTensor()
    ])

    mean = torch.zeros(3)

    std = torch.zeros(3)

    n = 0

    for path in tqdm(paths):

        try:

            img = Image.open(path).convert("RGB")

            img = preprocess_image(img)

            t = tfms(img)

            mean += t.mean(dim=[1,2])

            std += t.std(dim=[1,2])

            n += 1

        except:
            pass

    mean /= n

    std /= n

    return mean.tolist(), std.tolist()

MEAN, STD = compute_dataset_stats()

print("MEAN:", MEAN)
print("STD :", STD)

# ============================================================
# 9. TRANSFORMS
# ============================================================

train_tfms = transforms.Compose([

    transforms.Resize((IMG_SIZE, IMG_SIZE)),

    transforms.RandomHorizontalFlip(),

    transforms.RandomVerticalFlip(),

    transforms.RandomRotation(20),

    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),

    transforms.ToTensor(),

    transforms.Normalize(MEAN, STD)
])

val_tfms = transforms.Compose([

    transforms.Resize((IMG_SIZE, IMG_SIZE)),

    transforms.ToTensor(),

    transforms.Normalize(MEAN, STD)
])

# ============================================================
# 10. DATAFRAME
# ============================================================

def make_df(split):

    rows = []

    for c in CLASS_NAMES:

        folder = os.path.join(
            SPLIT_DATASET,
            split,
            c
        )

        for img in os.listdir(folder):

            rows.append({

                "path": os.path.join(folder,img),

                "label": c
            })

    return pd.DataFrame(rows)

train_df = make_df("train")

val_df = make_df("val")

test_df = make_df("test")

# ============================================================
# 11. DATASET CLASS
# ============================================================

class FungiDataset(Dataset):

    def __init__(self, df, tfms):

        self.df = df.reset_index(drop=True)

        self.tfms = tfms

    def __len__(self):

        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        img = Image.open(
            row["path"]
        ).convert("RGB")

        img = preprocess_image(img)

        img = self.tfms(img)

        label = label_map[row["label"]]

        return img, label

train_ds = FungiDataset(
    train_df,
    train_tfms
)

val_ds = FungiDataset(
    val_df,
    val_tfms
)

test_ds = FungiDataset(
    test_df,
    val_tfms
)

# ============================================================
# 12. WEIGHTED SAMPLER
# ============================================================

counts = train_df["label"].value_counts()

sample_weights = train_df["label"].map(
    lambda x: 1.0 / counts[x]
).values

sampler = WeightedRandomSampler(

    torch.DoubleTensor(sample_weights),

    len(sample_weights),

    replacement=True
)

# ============================================================
# 13. DATALOADER
# ============================================================

train_loader = DataLoader(

    train_ds,

    batch_size=BATCH_SIZE,

    sampler=sampler,

    num_workers=NUM_WORKERS,

    pin_memory=True
)

val_loader = DataLoader(

    val_ds,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=True
)

test_loader = DataLoader(

    test_ds,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=True
)

# ============================================================
# 14. CBAM
# ============================================================

class ChannelAttention(nn.Module):

    def __init__(self, channels, reduction=16):

        super().__init__()

        self.avg_pool = nn.AdaptiveAvgPool2d(1)

        self.max_pool = nn.AdaptiveMaxPool2d(1)

        self.fc = nn.Sequential(

            nn.Conv2d(
                channels,
                channels // reduction,
                1,
                bias=False
            ),

            nn.ReLU(),

            nn.Conv2d(
                channels // reduction,
                channels,
                1,
                bias=False
            )
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        avg_out = self.fc(self.avg_pool(x))

        max_out = self.fc(self.max_pool(x))

        out = avg_out + max_out

        return x * self.sigmoid(out)

class SpatialAttention(nn.Module):

    def __init__(self):

        super().__init__()

        self.conv = nn.Conv2d(
            2,
            1,
            kernel_size=7,
            padding=3
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        avg_out = torch.mean(
            x,
            dim=1,
            keepdim=True
        )

        max_out, _ = torch.max(
            x,
            dim=1,
            keepdim=True
        )

        x_cat = torch.cat(
            [avg_out,max_out],
            dim=1
        )

        return x * self.sigmoid(
            self.conv(x_cat)
        )

class CBAM(nn.Module):

    def __init__(self, channels):

        super().__init__()

        self.ca = ChannelAttention(channels)

        self.sa = SpatialAttention()

    def forward(self, x):

        x = self.ca(x)

        x = self.sa(x)

        return x

# ============================================================
# 15. PROPOSED MODEL
# ============================================================

class ProposedModel(nn.Module):

    def __init__(self, num_classes):

        super().__init__()

        self.backbone = timm.create_model(

            "tf_efficientnetv2_s",

            pretrained=True,

            num_classes=0,

            global_pool=""
        )

        features = self.backbone.num_features

        self.cbam = CBAM(features)

        self.pool = nn.AdaptiveAvgPool2d(1)

        self.classifier = nn.Sequential(

            nn.Flatten(),

            nn.Linear(features, 512),

            nn.BatchNorm1d(512),

            nn.GELU(),

            nn.Dropout(0.3),

            nn.Linear(512, num_classes)
        )

    def forward(self, x):

        x = self.backbone(x)

        x = self.cbam(x)

        x = self.pool(x)

        x = self.classifier(x)

        return x

# ============================================================
# 16. GENERIC BASELINE MODEL
# ============================================================

class GenericModel(nn.Module):

    def __init__(self, model_name, num_classes):

        super().__init__()

        self.backbone = timm.create_model(

            model_name,

            pretrained=True,

            num_classes=0
        )

        features = self.backbone.num_features

        self.classifier = nn.Sequential(

            nn.Linear(features, 512),

            nn.BatchNorm1d(512),

            nn.GELU(),

            nn.Dropout(0.3),

            nn.Linear(512, num_classes)
        )

    def forward(self, x):

        x = self.backbone(x)

        x = self.classifier(x)

        return x

# ============================================================
# 17. MODEL LIST
# ============================================================

MODEL_LIST = {

    "EfficientNetB0": "efficientnet_b0",

    "ResNet50": "resnet50",

    "DenseNet121": "densenet121",

    "MobileNetV3": "mobilenetv3_large_100",

    "ConvNeXtTiny": "convnext_tiny",

    "SwinTiny": "swin_tiny_patch4_window7_224",

    "Proposed_EfficientNetV2_CBAM": "proposed"
}

# ============================================================
# 18. TRAIN + EVALUATE
# ============================================================

results = []

for model_label, model_name in MODEL_LIST.items():

    print("\n" + "="*70)
    print("TRAINING:", model_label)
    print("="*70)

    if model_name == "proposed":

        model = ProposedModel(NUM_CLASSES)

    else:

        model = GenericModel(
            model_name,
            NUM_CLASSES
        )

    model = model.to(device)

    criterion = nn.CrossEntropyLoss(
        label_smoothing=0.1
    )

    optimizer = optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    scheduler = optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=EPOCHS
    )

    best_acc = 0

    patience_counter = 0

    train_losses = []

    val_losses = []

    train_accs = []

    val_accs = []

    # ========================================================
    # PARAMETERS + FLOPS
    # ========================================================

    dummy_input = torch.randn(
        1,
        3,
        IMG_SIZE,
        IMG_SIZE
    ).to(device)

    macs, params = profile(
        model,
        inputs=(dummy_input,),
        verbose=False
    )

    flops = macs * 2

    # ========================================================
    # TRAINING
    # ========================================================

    for epoch in range(EPOCHS):

        model.train()

        train_loss = 0

        correct = 0

        total = 0

        for images, labels in tqdm(train_loader):

            images = images.to(device)

            labels = labels.to(device)

            optimizer.zero_grad()

            outputs = model(images)

            loss = criterion(outputs, labels)

            loss.backward()

            optimizer.step()

            train_loss += loss.item()

            _, preds = torch.max(outputs, 1)

            correct += (preds == labels).sum().item()

            total += labels.size(0)

        train_acc = 100 * correct / total

        # ====================================================
        # VALIDATION
        # ====================================================

        model.eval()

        val_loss = 0

        correct = 0

        total = 0

        with torch.no_grad():

            for images, labels in val_loader:

                images = images.to(device)

                labels = labels.to(device)

                outputs = model(images)

                loss = criterion(outputs, labels)

                val_loss += loss.item()

                _, preds = torch.max(outputs,1)

                correct += (preds == labels).sum().item()

                total += labels.size(0)

        val_acc = 100 * correct / total

        scheduler.step()

        train_losses.append(
            train_loss / len(train_loader)
        )

        val_losses.append(
            val_loss / len(val_loader)
        )

        train_accs.append(train_acc)

        val_accs.append(val_acc)

        print(
            f"Epoch {epoch+1} | "
            f"Train Acc: {train_acc:.2f} | "
            f"Val Acc: {val_acc:.2f}"
        )

        if val_acc > best_acc:

            best_acc = val_acc

            torch.save(
                model.state_dict(),
                f"{model_label}.pth"
            )

            patience_counter = 0

        else:

            patience_counter += 1

        if patience_counter >= PATIENCE:

            print("Early Stopping")

            break

    # ========================================================
    # TRAINING CURVES
    # ========================================================

    plt.figure(figsize=(8,6))

    plt.plot(train_accs, label="Train")

    plt.plot(val_accs, label="Validation")

    plt.xlabel("Epoch")

    plt.ylabel("Accuracy")

    plt.title(f"{model_label} Accuracy Curve")

    plt.legend()

    plt.savefig(
        f"{model_label}_accuracy_curve.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    plt.figure(figsize=(8,6))

    plt.plot(train_losses, label="Train")

    plt.plot(val_losses, label="Validation")

    plt.xlabel("Epoch")

    plt.ylabel("Loss")

    plt.title(f"{model_label} Loss Curve")

    plt.legend()

    plt.savefig(
        f"{model_label}_loss_curve.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    # ========================================================
    # LOAD BEST MODEL
    # ========================================================

    model.load_state_dict(
        torch.load(f"{model_label}.pth")
    )

    # ========================================================
    # TESTING
    # ========================================================

    model.eval()

    y_true = []

    y_pred = []

    y_prob = []

    inference_times = []

    with torch.no_grad():

        for images, labels in tqdm(test_loader):

            images = images.to(device)

            start_time = time.time()

            outputs = model(images)

            end_time = time.time()

            inference_times.append(
                end_time - start_time
            )

            probs = torch.softmax(
                outputs,
                dim=1
            )

            preds = torch.argmax(
                probs,
                dim=1
            )

            y_true.extend(labels.numpy())

            y_pred.extend(preds.cpu().numpy())

            y_prob.extend(probs.cpu().numpy())

    y_true = np.array(y_true)

    y_pred = np.array(y_pred)

    y_prob = np.array(y_prob)

    # ========================================================
    # METRICS
    # ========================================================

    acc = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average='macro'
    )

    recall = recall_score(
        y_true,
        y_pred,
        average='macro'
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average='macro'
    )

    kappa = cohen_kappa_score(
        y_true,
        y_pred
    )

    mcc = matthews_corrcoef(
        y_true,
        y_pred
    )

    y_true_bin = label_binarize(
        y_true,
        classes=list(range(NUM_CLASSES))
    )

    macro_auc = roc_auc_score(

        y_true_bin,

        y_prob,

        multi_class='ovr',

        average='macro'
    )

    pr_auc_scores = []

    for i in range(NUM_CLASSES):

        precision_curve, recall_curve, _ = \
            precision_recall_curve(
                y_true_bin[:,i],
                y_prob[:,i]
            )

        pr_auc_scores.append(
            auc(recall_curve, precision_curve)
        )

    macro_pr_auc = np.mean(pr_auc_scores)

    inference_time = np.mean(inference_times)

    fps = BATCH_SIZE / inference_time

    # ========================================================
    # SAVE CLASSIFICATION REPORT
    # ========================================================

    report = classification_report(

        y_true,

        y_pred,

        target_names=CLASS_NAMES,

        output_dict=True
    )

    report_df = pd.DataFrame(report).transpose()

    report_df.to_csv(
        f"{model_label}_classification_report.csv"
    )

    # ========================================================
    # CONFUSION MATRIX
    # ========================================================

    cm = confusion_matrix(y_true, y_pred)

    plt.figure(figsize=(8,6))

    sns.heatmap(

        cm,

        annot=True,

        fmt='d',

        cmap='Blues',

        xticklabels=CLASS_NAMES,

        yticklabels=CLASS_NAMES
    )

    plt.xlabel("Predicted")

    plt.ylabel("True")

    plt.title(f"{model_label} Confusion Matrix")

    plt.savefig(

        f"{model_label}_confusion_matrix.png",

        dpi=300,

        bbox_inches='tight'
    )

    plt.close()

    # ========================================================
    # ROC CURVE
    # ========================================================

    plt.figure(figsize=(8,6))

    for i in range(NUM_CLASSES):

        fpr, tpr, _ = roc_curve(
            y_true_bin[:,i],
            y_prob[:,i]
        )

        roc_auc = auc(fpr,tpr)

        plt.plot(
            fpr,
            tpr,
            label=f"{CLASS_NAMES[i]} AUC={roc_auc:.3f}"
        )

    plt.plot([0,1],[0,1],'k--')

    plt.xlabel("False Positive Rate")

    plt.ylabel("True Positive Rate")

    plt.title(f"{model_label} ROC Curve")

    plt.legend()

    plt.savefig(

        f"{model_label}_roc_curve.png",

        dpi=300,

        bbox_inches='tight'
    )

    plt.close()

    # ========================================================
    # PR CURVE
    # ========================================================

    plt.figure(figsize=(8,6))

    for i in range(NUM_CLASSES):

        precision_curve, recall_curve, _ = \
            precision_recall_curve(
                y_true_bin[:,i],
                y_prob[:,i]
            )

        plt.plot(
            recall_curve,
            precision_curve,
            label=f"{CLASS_NAMES[i]}"
        )

    plt.xlabel("Recall")

    plt.ylabel("Precision")

    plt.title(f"{model_label} PR Curve")

    plt.legend()

    plt.savefig(

        f"{model_label}_pr_curve.png",

        dpi=300,

        bbox_inches='tight'
    )

    plt.close()

    # ========================================================
    # STORE RESULTS
    # ========================================================

    results.append({

        "Model": model_label,

        "Accuracy": round(acc*100,2),

        "Macro AUC": round(macro_auc,4),

        "Macro PR AUC": round(macro_pr_auc,4),

        "Precision": round(precision,4),

        "Recall": round(recall,4),

        "F1-Score": round(f1,4),

        "Kappa": round(kappa,4),

        "MCC": round(mcc,4),

        "Parameters(M)": round(params/1e6,2),

        "FLOPs(G)": round(flops/1e9,2),

        "Inference Time(ms)": round(
            inference_time*1000,
            2
        ),

        "FPS": round(fps,2),

        "Support": len(y_true)
    })

# ============================================================
# 19. FINAL COMPARISON TABLE
# ============================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
)

results_df.to_csv(

    os.path.join(
        SAVE_DIR,
        "all_model_comparison.csv"
    ),

    index=False
)

print("\n" + "="*80)
print("FINAL MODEL COMPARISON")
print("="*80)

print(results_df)

# ============================================================
# 20. BARPLOT COMPARISON
# ============================================================

plt.figure(figsize=(12,6))

sns.barplot(

    data=results_df,

    x="Model",

    y="Accuracy"
)

plt.xticks(rotation=20)

plt.ylabel("Accuracy (%)")

plt.title("Model Comparison Accuracy")

plt.savefig(

    os.path.join(
        SAVE_DIR,
        "model_comparison_accuracy.png"
    ),

    dpi=300,

    bbox_inches='tight'
)

plt.close()

# ============================================================
# 21. SAVE SUMMARY
# ============================================================

print("\n" + "="*80)
print("PIPELINE COMPLETE")
print("="*80)

print("\nSaved Files:")

print("all_model_comparison.csv")

print("classification reports")

print("confusion matrices")

print("ROC curves")

print("PR curves")

print("training curves")

print("loss curves")

print("best model weights")

print("barplot comparison")
```

```python
# ============================================================
# Q1-READY MICROSCOPIC FUNGI CLASSIFICATION PIPELINE
# MULTI-MODEL COMPARISON FRAMEWORK
# FINAL FIXED VERSION
# ============================================================

# ============================================================
# 1. INSTALL LIBRARIES
# ============================================================

!pip install timm opencv-python-headless seaborn thop -q

# ============================================================
# 2. IMPORTS
# ============================================================

import os
import cv2
import gc
import time
import random
import shutil
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader
from torch.utils.data import WeightedRandomSampler

from torchvision import transforms

import timm

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    roc_curve,
    auc,
    precision_recall_curve,
    cohen_kappa_score,
    matthews_corrcoef
)

from sklearn.preprocessing import label_binarize

from thop import profile

warnings.filterwarnings("ignore")

# ============================================================
# 3. CONFIGURATION
# ============================================================

ORIGINAL_DATASET = "/kaggle/input/datasets/joebeachcapital/defungi"

SPLIT_DATASET = "/kaggle/working/fungi_split_dataset"

SAVE_DIR = "/kaggle/working"

IMG_SIZE = 288

BATCH_SIZE = 32

EPOCHS = 20

LR = 2e-4

WEIGHT_DECAY = 1e-4

NUM_WORKERS = 2

PATIENCE = 5

SEED = 42

TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

# ============================================================
# 4. SEED
# ============================================================

def seed_everything(seed=42):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True

seed_everything(SEED)

# ============================================================
# 5. DEVICE
# ============================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("="*60)
print("DEVICE:", device)

if torch.cuda.is_available():

    print("GPU:", torch.cuda.get_device_name(0))

print("="*60)

# ============================================================
# 6. REMOVE OLD SPLIT
# ============================================================

if os.path.exists(SPLIT_DATASET):

    shutil.rmtree(SPLIT_DATASET)

# ============================================================
# 7. DATASET SPLITTING
# ============================================================

classes = ['H1', 'H2', 'H3', 'H5', 'H6']

for split in ['train', 'val', 'test']:

    for cls in classes:

        os.makedirs(

            os.path.join(
                SPLIT_DATASET,
                split,
                cls
            ),

            exist_ok=True
        )

for cls in classes:

    class_path = os.path.join(
        ORIGINAL_DATASET,
        cls
    )

    images = [

        img for img in os.listdir(class_path)

        if img.lower().endswith(
            ('.png','.jpg','.jpeg')
        )
    ]

    train_imgs, temp_imgs = train_test_split(

        images,

        test_size=(1 - TRAIN_RATIO),

        random_state=SEED,

        shuffle=True
    )

    val_imgs, test_imgs = train_test_split(

        temp_imgs,

        test_size=0.5,

        random_state=SEED,

        shuffle=True
    )

    for img in train_imgs:

        shutil.copy2(

            os.path.join(class_path, img),

            os.path.join(
                SPLIT_DATASET,
                'train',
                cls,
                img
            )
        )

    for img in val_imgs:

        shutil.copy2(

            os.path.join(class_path, img),

            os.path.join(
                SPLIT_DATASET,
                'val',
                cls,
                img
            )
        )

    for img in test_imgs:

        shutil.copy2(

            os.path.join(class_path, img),

            os.path.join(
                SPLIT_DATASET,
                'test',
                cls,
                img
            )
        )

print("\nDataset Split Complete")

# ============================================================
# 8. PREPROCESSING
# ============================================================

MORPH_KERNEL = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE,
    (3,3)
)

def preprocess_image(pil_img):

    img = cv2.cvtColor(
        np.array(pil_img),
        cv2.COLOR_RGB2BGR
    )

    img = cv2.medianBlur(img, 3)

    opened = cv2.morphologyEx(
        img,
        cv2.MORPH_OPEN,
        MORPH_KERNEL
    )

    closed = cv2.morphologyEx(
        opened,
        cv2.MORPH_CLOSE,
        MORPH_KERNEL
    )

    lab = cv2.cvtColor(
        closed,
        cv2.COLOR_BGR2LAB
    )

    l,a,b = cv2.split(lab)

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8,8)
    )

    l_eq = clahe.apply(l)

    merged = cv2.merge([l_eq,a,b])

    img = cv2.cvtColor(
        merged,
        cv2.COLOR_LAB2BGR
    )

    blurred = cv2.GaussianBlur(
        img,
        (0,0),
        sigmaX=1.5
    )

    img = cv2.addWeighted(
        img,
        1.5,
        blurred,
        -0.5,
        0
    )

    return Image.fromarray(
        cv2.cvtColor(
            img,
            cv2.COLOR_BGR2RGB
        )
    )

# ============================================================
# 9. DATASET STATS
# ============================================================

TRAIN_DIR = os.path.join(
    SPLIT_DATASET,
    "train"
)

CLASS_NAMES = sorted(
    os.listdir(TRAIN_DIR)
)

label_map = {
    c:i for i,c in enumerate(CLASS_NAMES)
}

NUM_CLASSES = len(CLASS_NAMES)

MEAN = [0.485, 0.456, 0.406]
STD = [0.229, 0.224, 0.225]

# ============================================================
# 10. TRANSFORMS
# ============================================================

train_tfms = transforms.Compose([

    transforms.Resize((IMG_SIZE, IMG_SIZE)),

    transforms.RandomHorizontalFlip(),

    transforms.RandomVerticalFlip(),

    transforms.RandomRotation(20),

    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),

    transforms.ToTensor(),

    transforms.Normalize(MEAN, STD)
])

val_tfms = transforms.Compose([

    transforms.Resize((IMG_SIZE, IMG_SIZE)),

    transforms.ToTensor(),

    transforms.Normalize(MEAN, STD)
])

# ============================================================
# 11. DATAFRAME
# ============================================================

def make_df(split):

    rows = []

    for c in CLASS_NAMES:

        folder = os.path.join(
            SPLIT_DATASET,
            split,
            c
        )

        for img in os.listdir(folder):

            rows.append({

                "path": os.path.join(folder,img),

                "label": c
            })

    return pd.DataFrame(rows)

train_df = make_df("train")

val_df = make_df("val")

test_df = make_df("test")

# ============================================================
# 12. DATASET CLASS
# ============================================================

class FungiDataset(Dataset):

    def __init__(self, df, tfms):

        self.df = df.reset_index(drop=True)

        self.tfms = tfms

    def __len__(self):

        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        img = Image.open(
            row["path"]
        ).convert("RGB")

        img = preprocess_image(img)

        img = self.tfms(img)

        label = label_map[row["label"]]

        return img, label

train_ds = FungiDataset(
    train_df,
    train_tfms
)

val_ds = FungiDataset(
    val_df,
    val_tfms
)

test_ds = FungiDataset(
    test_df,
    val_tfms
)

# ============================================================
# 13. WEIGHTED SAMPLER
# ============================================================

counts = train_df["label"].value_counts()

sample_weights = train_df["label"].map(
    lambda x: 1.0 / counts[x]
).values

sampler = WeightedRandomSampler(

    torch.DoubleTensor(sample_weights),

    len(sample_weights),

    replacement=True
)

# ============================================================
# 14. DATALOADER
# ============================================================

train_loader = DataLoader(

    train_ds,

    batch_size=BATCH_SIZE,

    sampler=sampler,

    num_workers=NUM_WORKERS,

    pin_memory=True
)

val_loader = DataLoader(

    val_ds,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=True
)

test_loader = DataLoader(

    test_ds,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=True
)

# ============================================================
# 15. CBAM
# ============================================================

class ChannelAttention(nn.Module):

    def __init__(self, channels, reduction=16):

        super().__init__()

        self.avg_pool = nn.AdaptiveAvgPool2d(1)

        self.max_pool = nn.AdaptiveMaxPool2d(1)

        self.fc = nn.Sequential(

            nn.Conv2d(
                channels,
                channels // reduction,
                1,
                bias=False
            ),

            nn.ReLU(),

            nn.Conv2d(
                channels // reduction,
                channels,
                1,
                bias=False
            )
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        avg_out = self.fc(self.avg_pool(x))

        max_out = self.fc(self.max_pool(x))

        return x * self.sigmoid(avg_out + max_out)

class SpatialAttention(nn.Module):

    def __init__(self):

        super().__init__()

        self.conv = nn.Conv2d(
            2,
            1,
            kernel_size=7,
            padding=3
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        avg_out = torch.mean(
            x,
            dim=1,
            keepdim=True
        )

        max_out, _ = torch.max(
            x,
            dim=1,
            keepdim=True
        )

        x_cat = torch.cat(
            [avg_out,max_out],
            dim=1
        )

        return x * self.sigmoid(
            self.conv(x_cat)
        )

class CBAM(nn.Module):

    def __init__(self, channels):

        super().__init__()

        self.ca = ChannelAttention(channels)

        self.sa = SpatialAttention()

    def forward(self, x):

        x = self.ca(x)

        x = self.sa(x)

        return x

# ============================================================
# 16. PROPOSED MODEL
# ============================================================

class ProposedModel(nn.Module):

    def __init__(self, num_classes):

        super().__init__()

        self.backbone = timm.create_model(

            "tf_efficientnetv2_s",

            pretrained=True,

            num_classes=0,

            global_pool='avg'
        )

        with torch.no_grad():

            dummy = torch.randn(
                1,
                3,
                IMG_SIZE,
                IMG_SIZE
            )

            features = self.backbone(dummy).shape[1]

        self.cbam = CBAM(features)

        self.classifier = nn.Sequential(

            nn.Linear(features, 512),

            nn.BatchNorm1d(512),

            nn.GELU(),

            nn.Dropout(0.3),

            nn.Linear(512, num_classes)
        )

    def forward(self, x):

        x = self.backbone.forward_features(x)

        x = self.cbam(x)

        x = F.adaptive_avg_pool2d(x, 1)

        x = torch.flatten(x, 1)

        x = self.classifier(x)

        return x

# ============================================================
# 17. GENERIC MODEL
# ============================================================

class GenericModel(nn.Module):

    def __init__(self, model_name, num_classes):

        super().__init__()

        self.backbone = timm.create_model(

            model_name,

            pretrained=True,

            num_classes=0,

            global_pool='avg'
        )

        with torch.no_grad():

            dummy = torch.randn(
                1,
                3,
                IMG_SIZE,
                IMG_SIZE
            )

            features = self.backbone(dummy).shape[1]

        self.classifier = nn.Sequential(

            nn.Linear(features, 512),

            nn.BatchNorm1d(512),

            nn.GELU(),

            nn.Dropout(0.3),

            nn.Linear(512, num_classes)
        )

    def forward(self, x):

        x = self.backbone(x)

        x = self.classifier(x)

        return x

# ============================================================
# 18. MODEL LIST
# ============================================================

MODEL_LIST = {

    "EfficientNetB0": "efficientnet_b0",

    "ResNet50": "resnet50",

    "DenseNet121": "densenet121",

    "MobileNetV3": "mobilenetv3_large_100",

    "ConvNeXtTiny": "convnext_tiny",

    "SwinTiny": "swin_tiny_patch4_window7_224",

    "Proposed_EfficientNetV2_CBAM": "proposed"
}

# ============================================================
# 19. TRAINING LOOP
# ============================================================

results = []

for model_label, model_name in MODEL_LIST.items():

    print("\n" + "="*70)
    print("TRAINING:", model_label)
    print("="*70)

    model_save_dir = os.path.join(
        SAVE_DIR,
        model_label
    )

    os.makedirs(
        model_save_dir,
        exist_ok=True
    )

    if model_name == "proposed":

        model = ProposedModel(NUM_CLASSES)

    else:

        model = GenericModel(
            model_name,
            NUM_CLASSES
        )

    model = model.to(device)

    scaler = torch.amp.GradScaler("cuda")

    criterion = nn.CrossEntropyLoss(
        label_smoothing=0.1
    )

    optimizer = optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY
    )

    scheduler = optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=EPOCHS
    )

    # ========================================================
    # FLOPS + PARAMS
    # ========================================================

    dummy_input = torch.randn(
        1,
        3,
        IMG_SIZE,
        IMG_SIZE
    ).to(device)

    try:

        macs, params = profile(
            model,
            inputs=(dummy_input,),
            verbose=False
        )

    except:

        macs = 0

        params = sum(
            p.numel()
            for p in model.parameters()
        )

    flops = macs * 2

    # ========================================================
    # TRAINING
    # ========================================================

    best_acc = 0

    patience_counter = 0

    train_accs = []
    val_accs = []

    train_losses = []
    val_losses = []

    for epoch in range(EPOCHS):

        model.train()

        running_loss = 0

        correct = 0

        total = 0

        for images, labels in tqdm(train_loader):

            images = images.to(device)

            labels = labels.to(device)

            optimizer.zero_grad()

            with torch.amp.autocast("cuda"):

                outputs = model(images)

                loss = criterion(
                    outputs,
                    labels
                )

            scaler.scale(loss).backward()

            scaler.step(optimizer)

            scaler.update()

            running_loss += loss.item()

            _, preds = torch.max(outputs,1)

            correct += (
                preds == labels
            ).sum().item()

            total += labels.size(0)

        train_acc = 100 * correct / total

        train_loss = (
            running_loss / len(train_loader)
        )

        # ====================================================
        # VALIDATION
        # ====================================================

        model.eval()

        val_loss = 0

        correct = 0

        total = 0

        with torch.no_grad():

            for images, labels in val_loader:

                images = images.to(device)

                labels = labels.to(device)

                outputs = model(images)

                loss = criterion(
                    outputs,
                    labels
                )

                val_loss += loss.item()

                _, preds = torch.max(outputs,1)

                correct += (
                    preds == labels
                ).sum().item()

                total += labels.size(0)

        val_acc = 100 * correct / total

        val_loss = val_loss / len(val_loader)

        scheduler.step()

        train_accs.append(train_acc)
        val_accs.append(val_acc)

        train_losses.append(train_loss)
        val_losses.append(val_loss)

        print(
            f"Epoch {epoch+1} | "
            f"Train Acc: {train_acc:.2f} | "
            f"Val Acc: {val_acc:.2f}"
        )

        if val_acc > best_acc:

            best_acc = val_acc

            torch.save(

                model.state_dict(),

                os.path.join(
                    model_save_dir,
                    "best_model.pth"
                )
            )

            patience_counter = 0

        else:

            patience_counter += 1

        if patience_counter >= PATIENCE:

            print("Early Stopping")

            break

    # ========================================================
    # CURVES
    # ========================================================

    plt.figure(figsize=(8,6))

    plt.plot(train_accs,label='Train')
    plt.plot(val_accs,label='Validation')

    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")

    plt.title(f"{model_label} Accuracy")

    plt.legend()

    plt.savefig(

        os.path.join(
            model_save_dir,
            "accuracy_curve.png"
        ),

        dpi=300,

        bbox_inches='tight'
    )

    plt.close()

    plt.figure(figsize=(8,6))

    plt.plot(train_losses,label='Train')
    plt.plot(val_losses,label='Validation')

    plt.xlabel("Epoch")
    plt.ylabel("Loss")

    plt.title(f"{model_label} Loss")

    plt.legend()

    plt.savefig(

        os.path.join(
            model_save_dir,
            "loss_curve.png"
        ),

        dpi=300,

        bbox_inches='tight'
    )

    plt.close()

    # ========================================================
    # LOAD BEST
    # ========================================================

    model.load_state_dict(

        torch.load(

            os.path.join(
                model_save_dir,
                "best_model.pth"
            )
        )
    )

    # ========================================================
    # TESTING
    # ========================================================

    model.eval()

    y_true = []
    y_pred = []
    y_prob = []

    inference_times = []

    with torch.no_grad():

        for images, labels in tqdm(test_loader):

            images = images.to(device)

            torch.cuda.synchronize()

            start_time = time.time()

            outputs = model(images)

            torch.cuda.synchronize()

            end_time = time.time()

            inference_times.append(
                end_time - start_time
            )

            probs = torch.softmax(
                outputs,
                dim=1
            )

            preds = torch.argmax(
                probs,
                dim=1
            )

            y_true.extend(labels.numpy())

            y_pred.extend(preds.cpu().numpy())

            y_prob.extend(probs.cpu().numpy())

    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    y_prob = np.array(y_prob)

    # ========================================================
    # METRICS
    # ========================================================

    acc = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        average='macro'
    )

    recall = recall_score(
        y_true,
        y_pred,
        average='macro'
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average='macro'
    )

    kappa = cohen_kappa_score(
        y_true,
        y_pred
    )

    mcc = matthews_corrcoef(
        y_true,
        y_pred
    )

    y_true_bin = label_binarize(
        y_true,
        classes=list(range(NUM_CLASSES))
    )

    try:

        macro_auc = roc_auc_score(

            y_true_bin,

            y_prob,

            multi_class='ovr',

            average='macro'
        )

    except:

        macro_auc = 0

    pr_auc_scores = []

    for i in range(NUM_CLASSES):

        try:

            precision_curve, recall_curve, _ = \
                precision_recall_curve(
                    y_true_bin[:,i],
                    y_prob[:,i]
                )

            pr_auc_scores.append(
                auc(
                    recall_curve,
                    precision_curve
                )
            )

        except:

            pr_auc_scores.append(0)

    macro_pr_auc = np.mean(pr_auc_scores)

    inference_time = np.mean(
        inference_times
    )

    fps = BATCH_SIZE / inference_time

    # ========================================================
    # CLASSIFICATION REPORT
    # ========================================================

    report = classification_report(

        y_true,

        y_pred,

        target_names=CLASS_NAMES,

        output_dict=True
    )

    pd.DataFrame(report).transpose().to_csv(

        os.path.join(
            model_save_dir,
            "classification_report.csv"
        )
    )

    # ========================================================
    # CONFUSION MATRIX
    # ========================================================

    cm = confusion_matrix(

        y_true,

        y_pred,

        normalize='true'
    )

    plt.figure(figsize=(8,6))

    sns.heatmap(

        cm,

        annot=True,

        fmt='.2f',

        cmap='Blues',

        xticklabels=CLASS_NAMES,

        yticklabels=CLASS_NAMES
    )

    plt.xlabel("Predicted")
    plt.ylabel("True")

    plt.title(
        f"{model_label} Confusion Matrix"
    )

    plt.savefig(

        os.path.join(
            model_save_dir,
            "confusion_matrix.png"
        ),

        dpi=300,

        bbox_inches='tight'
    )

    plt.close()

    # ========================================================
    # ROC CURVE
    # ========================================================

    plt.figure(figsize=(8,6))

    for i in range(NUM_CLASSES):

        fpr, tpr, _ = roc_curve(
            y_true_bin[:,i],
            y_prob[:,i]
        )

        roc_auc = auc(fpr,tpr)

        plt.plot(

            fpr,

            tpr,

            label=f"{CLASS_NAMES[i]} "
                  f"AUC={roc_auc:.3f}"
        )

    plt.plot([0,1],[0,1],'k--')

    plt.xlabel("False Positive Rate")

    plt.ylabel("True Positive Rate")

    plt.title(f"{model_label} ROC Curve")

    plt.legend()

    plt.savefig(

        os.path.join(
            model_save_dir,
            "roc_curve.png"
        ),

        dpi=300,

        bbox_inches='tight'
    )

    plt.close()

    # ========================================================
    # PR CURVE
    # ========================================================

    plt.figure(figsize=(8,6))

    for i in range(NUM_CLASSES):

        precision_curve, recall_curve, _ = \
            precision_recall_curve(
                y_true_bin[:,i],
                y_prob[:,i]
            )

        plt.plot(
            recall_curve,
            precision_curve,
            label=f"{CLASS_NAMES[i]}"
        )

    plt.xlabel("Recall")

    plt.ylabel("Precision")

    plt.title(f"{model_label} PR Curve")

    plt.legend()

    plt.savefig(

        os.path.join(
            model_save_dir,
            "pr_curve.png"
        ),

        dpi=300,

        bbox_inches='tight'
    )

    plt.close()

    # ========================================================
    # STORE RESULTS
    # ========================================================

    results.append({

        "Model": model_label,

        "Accuracy": round(acc*100,2),

        "Macro AUC": round(macro_auc,4),

        "Macro PR AUC": round(macro_pr_auc,4),

        "Precision": round(precision,4),

        "Recall": round(recall,4),

        "F1-Score": round(f1,4),

        "Kappa": round(kappa,4),

        "MCC": round(mcc,4),

        "Parameters(M)": round(params/1e6,2),

        "FLOPs(G)": round(flops/1e9,2),

        "Inference Time(ms)": round(
            inference_time*1000,
            2
        ),

        "FPS": round(fps,2),

        "Support": len(y_true)
    })

    del model

    gc.collect()

    torch.cuda.empty_cache()

# ============================================================
# 20. FINAL COMPARISON TABLE
# ============================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
)

results_df.to_csv(

    os.path.join(
        SAVE_DIR,
        "all_model_comparison.csv"
    ),

    index=False
)

print("\n" + "="*80)
print("FINAL MODEL COMPARISON")
print("="*80)

print(results_df)

# ============================================================
# 21. BARPLOT
# ============================================================

plt.figure(figsize=(12,6))

sns.barplot(

    data=results_df,

    x="Model",

    y="Accuracy"
)

plt.xticks(rotation=20)

plt.ylabel("Accuracy (%)")

plt.title("Model Comparison Accuracy")

plt.savefig(

    os.path.join(
        SAVE_DIR,
        "model_comparison_accuracy.png"
    ),

    dpi=300,

    bbox_inches='tight'
)

plt.close()

# ============================================================
# 22. FINISHED
# ============================================================

print("\n" + "="*80)
print("PIPELINE COMPLETE")
print("="*80)

print("\nSaved Files:")
print("all_model_comparison.csv")
print("classification reports")
print("confusion matrices")
print("ROC curves")
print("PR curves")
print("training curves")
print("loss curves")
print("best model weights")
print("comparison barplot")
```

```python
# ============================================================
# Q1-READY MICROSCOPIC FUNGI CLASSIFICATION PIPELINE
# MULTI-MODEL COMPARISON FRAMEWORK
# ONE-BY-ONE MODEL TRAINING + COMPLETE METRICS
# ============================================================

# ============================================================
# 1. INSTALL LIBRARIES
# ============================================================

!pip install timm opencv-python-headless seaborn thop -q

# ============================================================
# 2. IMPORTS
# ============================================================

import os
import gc
import cv2
import time
import timm
import random
import shutil
import warnings
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from PIL import Image
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.optim as optim

from torchvision import transforms
from torch.utils.data import (
    Dataset,
    DataLoader,
    WeightedRandomSampler
)

from sklearn.model_selection import train_test_split

from sklearn.metrics import (

    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    roc_curve,
    auc,
    precision_recall_curve,
    cohen_kappa_score,
    matthews_corrcoef
)

from sklearn.preprocessing import label_binarize

from thop import profile

warnings.filterwarnings("ignore")

# ============================================================
# 3. CONFIGURATION
# ============================================================

ORIGINAL_DATASET = "/kaggle/input/datasets/joebeachcapital/defungi"

SPLIT_DATASET = "/kaggle/working/fungi_split_dataset"

SAVE_DIR = "/kaggle/working"

IMG_SIZE = 224

BATCH_SIZE = 32

EPOCHS = 20

LR = 2e-4

WEIGHT_DECAY = 1e-4

PATIENCE = 5

NUM_WORKERS = 2

SEED = 42

TRAIN_RATIO = 0.70

VAL_RATIO = 0.15

TEST_RATIO = 0.15

# ============================================================
# 4. SEED
# ============================================================

def seed_everything(seed=42):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True

seed_everything(SEED)

# ============================================================
# 5. DEVICE
# ============================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("="*60)
print("DEVICE:", device)

if torch.cuda.is_available():

    print("GPU:", torch.cuda.get_device_name(0))

print("="*60)

# ============================================================
# 6. DATASET SPLIT
# ============================================================

classes = ['H1', 'H2', 'H3', 'H5', 'H6']

if not os.path.exists(SPLIT_DATASET):

    for split in ['train', 'val', 'test']:

        for cls in classes:

            os.makedirs(
                os.path.join(
                    SPLIT_DATASET,
                    split,
                    cls
                ),
                exist_ok=True
            )

    for cls in classes:

        class_path = os.path.join(
            ORIGINAL_DATASET,
            cls
        )

        images = [

            img for img in os.listdir(class_path)

            if img.lower().endswith(
                ('.png','.jpg','.jpeg')
            )
        ]

        train_imgs, temp_imgs = train_test_split(

            images,

            test_size=(1 - TRAIN_RATIO),

            random_state=SEED,

            shuffle=True
        )

        val_imgs, test_imgs = train_test_split(

            temp_imgs,

            test_size=0.5,

            random_state=SEED,

            shuffle=True
        )

        for img in train_imgs:

            shutil.copy2(

                os.path.join(class_path, img),

                os.path.join(
                    SPLIT_DATASET,
                    'train',
                    cls,
                    img
                )
            )

        for img in val_imgs:

            shutil.copy2(

                os.path.join(class_path, img),

                os.path.join(
                    SPLIT_DATASET,
                    'val',
                    cls,
                    img
                )
            )

        for img in test_imgs:

            shutil.copy2(

                os.path.join(class_path, img),

                os.path.join(
                    SPLIT_DATASET,
                    'test',
                    cls,
                    img
                )
            )

print("\nDataset Split Complete")

# ============================================================
# 7. PREPROCESSING
# ============================================================

MORPH_KERNEL = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE,
    (3,3)
)

def preprocess_image(pil_img):

    img = cv2.cvtColor(
        np.array(pil_img),
        cv2.COLOR_RGB2BGR
    )

    img = cv2.medianBlur(img, 3)

    img = cv2.morphologyEx(
        img,
        cv2.MORPH_OPEN,
        MORPH_KERNEL
    )

    lab = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2LAB
    )

    l,a,b = cv2.split(lab)

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8,8)
    )

    l = clahe.apply(l)

    lab = cv2.merge([l,a,b])

    img = cv2.cvtColor(
        lab,
        cv2.COLOR_LAB2BGR
    )

    blur = cv2.GaussianBlur(
        img,
        (0,0),
        sigmaX=1.5
    )

    img = cv2.addWeighted(
        img,
        1.5,
        blur,
        -0.5,
        0
    )

    return Image.fromarray(
        cv2.cvtColor(
            img,
            cv2.COLOR_BGR2RGB
        )
    )

# ============================================================
# 8. DATASET STATS
# ============================================================

TRAIN_DIR = os.path.join(
    SPLIT_DATASET,
    "train"
)

CLASS_NAMES = sorted(
    os.listdir(TRAIN_DIR)
)

NUM_CLASSES = len(CLASS_NAMES)

label_map = {
    c:i for i,c in enumerate(CLASS_NAMES)
}

MEAN = [0.485, 0.456, 0.406]

STD = [0.229, 0.224, 0.225]

# ============================================================
# 9. TRANSFORMS
# ============================================================

train_tfms = transforms.Compose([

    transforms.Resize((IMG_SIZE,IMG_SIZE)),

    transforms.RandomHorizontalFlip(),

    transforms.RandomVerticalFlip(),

    transforms.RandomRotation(20),

    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),

    transforms.ToTensor(),

    transforms.Normalize(MEAN,STD)
])

val_tfms = transforms.Compose([

    transforms.Resize((IMG_SIZE,IMG_SIZE)),

    transforms.ToTensor(),

    transforms.Normalize(MEAN,STD)
])

# ============================================================
# 10. DATAFRAME
# ============================================================

def make_df(split):

    rows = []

    for c in CLASS_NAMES:

        folder = os.path.join(
            SPLIT_DATASET,
            split,
            c
        )

        for img in os.listdir(folder):

            rows.append({

                "path": os.path.join(folder,img),

                "label": c
            })

    return pd.DataFrame(rows)

train_df = make_df("train")

val_df = make_df("val")

test_df = make_df("test")

# ============================================================
# 11. DATASET CLASS
# ============================================================

class FungiDataset(Dataset):

    def __init__(self, df, tfms):

        self.df = df.reset_index(drop=True)

        self.tfms = tfms

    def __len__(self):

        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        img = Image.open(
            row["path"]
        ).convert("RGB")

        img = preprocess_image(img)

        img = self.tfms(img)

        label = label_map[row["label"]]

        return img, label

train_ds = FungiDataset(train_df, train_tfms)

val_ds = FungiDataset(val_df, val_tfms)

test_ds = FungiDataset(test_df, val_tfms)

# ============================================================
# 12. SAMPLER
# ============================================================

counts = train_df["label"].value_counts()

sample_weights = train_df["label"].map(
    lambda x: 1.0 / counts[x]
).values

sampler = WeightedRandomSampler(

    torch.DoubleTensor(sample_weights),

    len(sample_weights),

    replacement=True
)

# ============================================================
# 13. DATALOADER
# ============================================================

train_loader = DataLoader(

    train_ds,

    batch_size=BATCH_SIZE,

    sampler=sampler,

    num_workers=NUM_WORKERS,

    pin_memory=True
)

val_loader = DataLoader(

    val_ds,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=True
)

test_loader = DataLoader(

    test_ds,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=NUM_WORKERS,

    pin_memory=True
)

# ============================================================
# 14. CBAM
# ============================================================

class ChannelAttention(nn.Module):

    def __init__(self, channels, reduction=16):

        super().__init__()

        self.avg_pool = nn.AdaptiveAvgPool2d(1)

        self.max_pool = nn.AdaptiveMaxPool2d(1)

        self.fc = nn.Sequential(

            nn.Conv2d(
                channels,
                channels // reduction,
                1,
                bias=False
            ),

            nn.ReLU(),

            nn.Conv2d(
                channels // reduction,
                channels,
                1,
                bias=False
            )
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        avg_out = self.fc(self.avg_pool(x))

        max_out = self.fc(self.max_pool(x))

        out = avg_out + max_out

        return x * self.sigmoid(out)

class SpatialAttention(nn.Module):

    def __init__(self):

        super().__init__()

        self.conv = nn.Conv2d(
            2,
            1,
            kernel_size=7,
            padding=3
        )

        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        avg_out = torch.mean(
            x,
            dim=1,
            keepdim=True
        )

        max_out, _ = torch.max(
            x,
            dim=1,
            keepdim=True
        )

        x_cat = torch.cat(
            [avg_out,max_out],
            dim=1
        )

        return x * self.sigmoid(
            self.conv(x_cat)
        )

class CBAM(nn.Module):

    def __init__(self, channels):

        super().__init__()

        self.ca = ChannelAttention(channels)

        self.sa = SpatialAttention()

    def forward(self, x):

        x = self.ca(x)

        x = self.sa(x)

        return x

# ============================================================
# 15. MODELS
# ============================================================

class ProposedModel(nn.Module):

    def __init__(self, num_classes):

        super().__init__()

        self.backbone = timm.create_model(

            "tf_efficientnetv2_s",

            pretrained=True,

            num_classes=0,

            global_pool=""
        )

        features = self.backbone.num_features

        self.cbam = CBAM(features)

        self.pool = nn.AdaptiveAvgPool2d(1)

        self.classifier = nn.Sequential(

            nn.Flatten(),

            nn.Linear(features,512),

            nn.BatchNorm1d(512),

            nn.GELU(),

            nn.Dropout(0.3),

            nn.Linear(512,num_classes)
        )

    def forward(self, x):

        x = self.backbone(x)

        x = self.cbam(x)

        x = self.pool(x)

        x = self.classifier(x)

        return x

class GenericModel(nn.Module):

    def __init__(self, model_name, num_classes):

        super().__init__()

        self.backbone = timm.create_model(

            model_name,

            pretrained=True,

            num_classes=num_classes
        )

    def forward(self, x):

        return self.backbone(x)

# ============================================================
# 16. MODEL LIST
# ============================================================

MODEL_LIST = {

    "ResNet50": "resnet50",

    "DenseNet121": "densenet121",

    "MobileNetV3": "mobilenetv3_large_100",

    "EfficientNetB0": "efficientnet_b0",

    "ConvNeXtTiny": "convnext_tiny",

    "SwinTiny": "swin_tiny_patch4_window7_224",

    "Proposed_EfficientNetV2_CBAM": "proposed"
}

# ============================================================
# 17. RESULTS STORAGE
# ============================================================

results = []

# ============================================================
# 18. TRAIN MODELS ONE-BY-ONE
# ============================================================

for model_label, model_name in MODEL_LIST.items():

    print("\n" + "="*80)
    print(f"TRAINING MODEL: {model_label}")
    print("="*80)

    model_save_dir = os.path.join(
        SAVE_DIR,
        model_label
    )

    os.makedirs(model_save_dir, exist_ok=True)

    if model_name == "proposed":

        model = ProposedModel(NUM_CLASSES)

    else:

        model = GenericModel(
            model_name,
            NUM_CLASSES
        )

    model = model.to(device)

    # ========================================================
    # FLOPS
    # ========================================================

    dummy_input = torch.randn(
        1,
        3,
        IMG_SIZE,
        IMG_SIZE
    ).to(device)

    try:

        macs, params = profile(
            model,
            inputs=(dummy_input,),
            verbose=False
        )

        flops = macs * 2

    except:

        params = sum(
            p.numel() for p in model.parameters()
        )

        flops = 0

    # ========================================================
    # OPTIMIZER
    # ========================================================

    criterion = nn.CrossEntropyLoss(
        label_smoothing=0.1
    )

    optimizer = optim.AdamW(

        model.parameters(),

        lr=LR,

        weight_decay=WEIGHT_DECAY
    )

    scheduler = optim.lr_scheduler.CosineAnnealingLR(

        optimizer,

        T_max=EPOCHS
    )

    # ========================================================
    # TRAINING
    # ========================================================

    best_acc = 0

    patience_counter = 0

    train_losses = []

    val_losses = []

    train_accs = []

    val_accs = []

    for epoch in range(EPOCHS):

        print(f"\nEpoch [{epoch+1}/{EPOCHS}]")

        # ====================================================
        # TRAIN
        # ====================================================

        model.train()

        train_loss = 0

        correct = 0

        total = 0

        for images, labels in tqdm(train_loader):

            images = images.to(device)

            labels = labels.to(device)

            optimizer.zero_grad()

            outputs = model(images)

            loss = criterion(outputs, labels)

            loss.backward()

            optimizer.step()

            train_loss += loss.item()

            _, preds = torch.max(outputs,1)

            correct += (preds == labels).sum().item()

            total += labels.size(0)

        train_acc = 100 * correct / total

        # ====================================================
        # VALIDATION
        # ====================================================

        model.eval()

        val_loss = 0

        correct = 0

        total = 0

        with torch.no_grad():

            for images, labels in val_loader:

                images = images.to(device)

                labels = labels.to(device)

                outputs = model(images)

                loss = criterion(outputs, labels)

                val_loss += loss.item()

                _, preds = torch.max(outputs,1)

                correct += (preds == labels).sum().item()

                total += labels.size(0)

        val_acc = 100 * correct / total

        scheduler.step()

        train_losses.append(
            train_loss / len(train_loader)
        )

        val_losses.append(
            val_loss / len(val_loader)
        )

        train_accs.append(train_acc)

        val_accs.append(val_acc)

        print(
            f"Train Acc: {train_acc:.2f}% | "
            f"Val Acc: {val_acc:.2f}%"
        )

        if val_acc > best_acc:

            best_acc = val_acc

            torch.save(

                model.state_dict(),

                os.path.join(
                    model_save_dir,
                    "best_model.pth"
                )
            )

            patience_counter = 0

        else:

            patience_counter += 1

        if patience_counter >= PATIENCE:

            print("Early Stopping")

            break

    # ========================================================
    # LOAD BEST MODEL
    # ========================================================

    model.load_state_dict(

        torch.load(
            os.path.join(
                model_save_dir,
                "best_model.pth"
            )
        )
    )

    # ========================================================
    # TESTING
    # ========================================================

    model.eval()

    y_true = []

    y_pred = []

    y_prob = []

    inference_times = []

    with torch.no_grad():

        for images, labels in tqdm(test_loader):

            images = images.to(device)

            start_time = time.time()

            outputs = model(images)

            end_time = time.time()

            inference_times.append(
                end_time - start_time
            )

            probs = torch.softmax(
                outputs,
                dim=1
            )

            preds = torch.argmax(
                probs,
                dim=1
            )

            y_true.extend(labels.numpy())

            y_pred.extend(preds.cpu().numpy())

            y_prob.extend(probs.cpu().numpy())

    y_true = np.array(y_true)

    y_pred = np.array(y_pred)

    y_prob = np.array(y_prob)

    # ========================================================
    # METRICS
    # ========================================================

    acc = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average='macro'
    )

    recall = recall_score(
        y_true,
        y_pred,
        average='macro'
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average='macro'
    )

    kappa = cohen_kappa_score(
        y_true,
        y_pred
    )

    mcc = matthews_corrcoef(
        y_true,
        y_pred
    )

    y_true_bin = label_binarize(
        y_true,
        classes=list(range(NUM_CLASSES))
    )

    macro_auc = roc_auc_score(

        y_true_bin,

        y_prob,

        multi_class='ovr',

        average='macro'
    )

    pr_auc_scores = []

    for i in range(NUM_CLASSES):

        precision_curve, recall_curve, _ = \
            precision_recall_curve(
                y_true_bin[:,i],
                y_prob[:,i]
            )

        pr_auc_scores.append(
            auc(recall_curve, precision_curve)
        )

    macro_pr_auc = np.mean(pr_auc_scores)

    cm_raw = confusion_matrix(
        y_true,
        y_pred
    )

    sensitivity_per_class = []

    specificity_per_class = []

    for i in range(NUM_CLASSES):

        TP = cm_raw[i, i]

        FN = np.sum(cm_raw[i, :]) - TP

        FP = np.sum(cm_raw[:, i]) - TP

        TN = np.sum(cm_raw) - (TP + FN + FP)

        sensitivity = TP / (TP + FN + 1e-8)

        specificity = TN / (TN + FP + 1e-8)

        sensitivity_per_class.append(sensitivity)

        specificity_per_class.append(specificity)

    macro_sensitivity = np.mean(
        sensitivity_per_class
    )

    macro_specificity = np.mean(
        specificity_per_class
    )

    inference_time = np.mean(
        inference_times
    )

    fps = BATCH_SIZE / inference_time

    # ========================================================
    # PRINT RESULTS
    # ========================================================

    print("\n" + "="*80)
    print(f"FINAL RESULTS: {model_label}")
    print("="*80)

    print(f"Accuracy           : {acc*100:.2f}%")
    print(f"Precision          : {precision:.4f}")
    print(f"Recall             : {recall:.4f}")
    print(f"F1-Score           : {f1:.4f}")
    print(f"Macro AUC          : {macro_auc:.4f}")
    print(f"Macro PR AUC       : {macro_pr_auc:.4f}")
    print(f"Kappa              : {kappa:.4f}")
    print(f"MCC                : {mcc:.4f}")
    print(f"Sensitivity        : {macro_sensitivity:.4f}")
    print(f"Specificity        : {macro_specificity:.4f}")

    print("\nClassification Report:\n")

    print(

        classification_report(
            y_true,
            y_pred,
            target_names=CLASS_NAMES
        )
    )

    # ========================================================
    # SAVE RESULTS
    # ========================================================

    results.append({

        "Model": model_label,

        "Accuracy": round(acc*100,2),

        "Macro AUC": round(macro_auc,4),

        "Macro PR AUC": round(macro_pr_auc,4),

        "Precision": round(precision,4),

        "Recall": round(recall,4),

        "F1-Score": round(f1,4),

        "Sensitivity": round(macro_sensitivity,4),

        "Specificity": round(macro_specificity,4),

        "Kappa": round(kappa,4),

        "MCC": round(mcc,4),

        "Parameters(M)": round(params/1e6,2),

        "FLOPs(G)": round(flops/1e9,2),

        "Inference(ms)": round(
            inference_time*1000,
            2
        ),

        "FPS": round(fps,2)
    })

    # ========================================================
    # CONFUSION MATRIX
    # ========================================================

    plt.figure(figsize=(8,6))

    sns.heatmap(

        cm_raw,

        annot=True,

        fmt='d',

        cmap='Blues',

        xticklabels=CLASS_NAMES,

        yticklabels=CLASS_NAMES
    )

    plt.xlabel("Predicted")

    plt.ylabel("True")

    plt.title(f"{model_label} Confusion Matrix")

    plt.savefig(

        os.path.join(
            model_save_dir,
            "confusion_matrix.png"
        ),

        dpi=300,

        bbox_inches='tight'
    )

    plt.close()

    # ========================================================
    # ROC CURVE
    # ========================================================

    plt.figure(figsize=(8,6))

    for i in range(NUM_CLASSES):

        fpr, tpr, _ = roc_curve(
            y_true_bin[:,i],
            y_prob[:,i]
        )

        roc_auc = auc(fpr,tpr)

        plt.plot(
            fpr,
            tpr,
            label=f"{CLASS_NAMES[i]} AUC={roc_auc:.3f}"
        )

    plt.plot([0,1],[0,1],'k--')

    plt.xlabel("False Positive Rate")

    plt.ylabel("True Positive Rate")

    plt.title(f"{model_label} ROC Curve")

    plt.legend()

    plt.savefig(

        os.path.join(
            model_save_dir,
            "roc_curve.png"
        ),

        dpi=300,

        bbox_inches='tight'
    )

    plt.close()

    # ========================================================
    # CLEAN MEMORY
    # ========================================================

    del model

    gc.collect()

    torch.cuda.empty_cache()

# ============================================================
# FINAL COMPARISON TABLE
# ============================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
)

results_df.to_csv(

    os.path.join(
        SAVE_DIR,
        "all_model_comparison.csv"
    ),

    index=False
)

print("\n" + "="*80)
print("FINAL MODEL COMPARISON")
print("="*80)

print(results_df)
```
